** Verificación de conexión de base de datos

```python
import duckdb, pandas as pd

# Uso la ruta absoluta para que funcione sin importar la carpeta del notebook
RUTA_DB = "/workspaces/prueba-tecnica-cxc/data/fuente_cxc.sqlite"  # <-- pon aquí la ruta que te imprimió

con = duckdb.connect()
con.execute("INSTALL sqlite; LOAD sqlite;")
con.execute(f"ATTACH '{RUTA_DB}' AS fuente (TYPE sqlite);")

print(con.execute("SELECT COUNT(*) AS filas FROM fuente.tabla1").df())
con.execute("SELECT * FROM fuente.tabla1 LIMIT 5").df()
```
